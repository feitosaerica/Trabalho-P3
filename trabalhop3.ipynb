```python
# Cálculo de Desconto por Quantidade em Cada Ano

# 1. Informar a quantidade de anos, o ano inicial e o valor unitário
anos = int(input("Digite a quantidade de anos que deseja simular: "))
ano_inicial = int(input("Digite o ano inicial da simulação (ex: 2024): "))
valor_unitario = float(input("Digite o valor unitário do produto: ").replace(',', '.'))

# Listas para armazenar os dados de cada ano
nome_dos_anos = []
quantidades_por_ano = []
valores_sem_desconto_por_ano = []
porcentagens_desconto_por_ano = []
valores_desconto_por_ano = []
valores_com_desconto_por_ano = []

# 2. Informar a quantidade individual para cada ano sequencial e calcular o desconto individualmente
for i in range(anos):
    ano_atual = ano_inicial + i
    nome_dos_anos.append(str(ano_atual))

    qtd_ano = float(input(f"Digite a quantidade comprada em {ano_atual}: ").replace(',', '.'))
    quantidades_por_ano.append(qtd_ano)

    # Calcular valor sem desconto do ano
    valor_sem_desc = qtd_ano * valor_unitario
    valores_sem_desconto_por_ano.append(valor_sem_desc)

    # Definir a porcentagem de desconto com base na quantidade DO ANO
    if qtd_ano < 100:
        pct_desc = 0.0
    elif 100 <= qtd_ano < 500:
        pct_desc = 5.0
    elif 500 <= qtd_ano < 1000:
        pct_desc = 10.0
    else:
        pct_desc = 15.0

    porcentagens_desconto_por_ano.append(pct_desc)

    # Calcular descontos e valor final do ano
    v_desc = valor_sem_desc * (pct_desc / 100)
    valores_desconto_por_ano.append(v_desc)

    v_com_desc = valor_sem_desc - v_desc
    valores_com_desconto_por_ano.append(v_com_desc)

# 3. Exibir o resultado detalhado por ano e o acumulado final
print("\n================ RESULTADOS POR ANO ================")
for i in range(anos):
    print(f"\n--- Ano: {nome_dos_anos[i]} ---")
    print(f"  Quantidade comprada: {quantidades_por_ano[i]:.2f}")
    print(f"  Valor sem desconto:  R$ {valores_sem_desconto_por_ano[i]:.2f}")
    print(f"  Desconto aplicado:   {porcentagens_desconto_por_ano[i]:.1f}% (R$ {valores_desconto_por_ano[i]:.2f})")
    print(f"  Valor com desconto:  R$ {valores_com_desconto_por_ano[i]:.2f}")

print("\n================ RESUMO ACUMULADO ================")
print(f"Quantidade total acumulada: {sum(quantidades_por_ano):.2f}")
print(f"Valor total sem desconto: R$ {sum(valores_sem_desconto_por_ano):.2f}")
print(f"Valor total economizado (descontos): R$ {sum(valores_desconto_por_ano):.2f}")
print(f"Valor total final pago: R$ {sum(valores_com_desconto_por_ano):.2f}")
```
