```python
# Cálculo de Desconto por Quantidade Anual

# 1. Informar a quantidade comprada no ano e o valor unitário (ou total sem desconto)
qtd_anual = float(input("Digite a quantidade comprada no ano: "))
valor_unitario = float(input("Digite o valor unitário do produto: "))

valor_total_sem_desconto = qtd_anual * valor_unitario

# 2. Definir a porcentagem de desconto com base na quantidade
# Ajuste os limites (valores) e as taxas conforme a sua necessidade
if qtd_anual < 100:
    porcentagem_desconto = 0.0  # Sem desconto
elif qtd_anual >= 100 and qtd_anual < 500:
    porcentagem_desconto = 5.0  # 5% de desconto
elif qtd_anual >= 500 and qtd_anual < 1000:
    porcentagem_desconto = 10.0 # 10% de desconto
else:
    porcentagem_desconto = 15.0 # 15% de desconto para 1000 ou mais

# 3. Calcular os valores finais
valor_desconto = valor_total_sem_desconto * (porcentagem_desconto / 100)
valor_com_desconto = valor_total_sem_desconto - valor_desconto

# 4. Exibir o resultado
print(f"\n--- Resumo da Compra ---")
print(f"Quantidade anual: {qtd_anual:.2f}")
print(f"Valor total sem desconto: R$ {valor_total_sem_desconto:.2f}")
print(f"Desconto aplicado: {porcentagem_desconto}% (R$ {valor_desconto:.2f})")
print(f"Valor final com desconto: R$ {valor_com_desconto:.2f}")
```
